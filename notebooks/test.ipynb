```python
def clean_load(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()
    df["period"] = pd.to_datetime(df["period"], format="%Y-%m-%dT%H", utc=True)
    df = df.set_index("period")
    df.index.name = "date"
    #df.index.freq = "h"
    return df
```
